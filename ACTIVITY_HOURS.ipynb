```
#program 1: Calculate total weekly energy from a list
daily_kwh = [400, 850, 500, 300, 600, 700, 450, 560, 780, 690 ]
total_energy = sum(daily_kwh)
print("Total Energy Used:", total_energy, "kwh")
```

```
#program 2: store building usage data
building_usage = {
    "CSE": 400,
    "HOSTEL": 300
}
print("Hostel used:", building_usage["HOSTEL"], "kwh")
```

```
#program 3:A function to calculate electricity cost
def calculate_bill(kwh, rate_per_unit):
    return kwh * rate_per_unit
todays_bill = calculate_bill(850, 8)
print("Today's estimated bill is Rs.", todays_bill)
```

```
#import the pandas library
import pandas as pd

#Load the dataset you created
df = pd.read_csv('campus_energy.csv')

#Display the first few rows of the data
print("Here is the raw data:")
display(df.head())

#Group the data to see total usage per building
building_totals = df.groupby('Building')['kwh-used'].sum()
print("\nTotal usage per building:")
print(building_totals)

#Generate a statistical summary (average, min, max) of the numeric columns
print("\nStatistical Summary:")
display(df.describe())
```
